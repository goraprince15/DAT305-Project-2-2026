## 1. Introduction and Research Question

### Background
Offshore oil wells operate under complex conditions, where abnormal events can lead to production losses, increased maintenance costs, environmental damage, and safety risks. Early detection of these events is therefore important for improving operational reliability. Petrobras developed the 3W dataset to support research into machine-learning methods for detecting and classifying abnormal conditions using oil-well production data.

Although machine-learning models have already been investigated for this purpose, detecting faults promptly while avoiding unnecessary false alarms remains an important consideration. This project investigates whether changing the length of historical sensor data examined by a Random Forest model affects fault-detection delay and false-alarm frequency.

### Research Question
How does sliding-window length affect the trade-off between fault-detection delay and false alarms when using Random Forest on the Petrobras 3W dataset?

### Research Objective
The main objective of this study is to investigate how different sliding-window lengths affect the performance of a Random Forest model in detecting abnormal operating conditions in oil wells using the Petrobras 3W dataset.

Specifically, the study aims to:

1. Vary the sliding-window length to examine how different amounts of historical sensor data influence anomaly detection.

2. Measure fault-detection delay and the frequency of false alarms for each window length.

3. Compare the results to determine which window length provides the most suitable balance between timely fault detection and minimising false alarms.
